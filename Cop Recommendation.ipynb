```python
#Handle outliers

# Selecting numerical columns
numeric_cols = ['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall']

def cap_outliers(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    return series.clip(lower_bound, upper_bound)


for col in numeric_cols:

    # Calculate IQR boundaries
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    # Count outliers before capping
    outliers_before = ((df[col] < lower) | (df[col] > upper)).sum()

    # Cap outliers
    df[col] = cap_outliers(df[col])

    # Check outliers after capping
    q1_new = df[col].quantile(0.25)
    q3_new = df[col].quantile(0.75)
    iqr_new = q3_new - q1_new

    lower_new = q1_new - 1.5 * iqr_new
    upper_new = q3_new + 1.5 * iqr_new

    outliers_after = (
        (df[col] < lower_new) | (df[col] > upper_new)
    ).sum()

    print(
        f"{col}: {outliers_before} outliers capped → "
        f"{outliers_after} remaining"
    )
```
